```python
type_counts = df['type'].value_counts()

plt.figure(figsize=(8,5))
plt.bar(type_counts.index, type_counts.values)

plt.title("Distribution of Movies and TV Shows")
plt.xlabel("Content Type")
plt.ylabel("Number of Titles")

plt.show()
```

```python
year_counts = df['release_year'].value_counts().sort_index()

plt.figure(figsize=(12,6))
plt.plot(year_counts.index, year_counts.values, marker='o')

plt.title("Content Released by Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")

plt.grid(True)
plt.show()
```

```python
genre_counts = df['genres'].value_counts().head(10)

plt.figure(figsize=(10,6))
plt.barh(genre_counts.index[::-1], genre_counts.values[::-1])

plt.title("Top 10 Content Genres")
plt.xlabel("Number of Titles")
plt.ylabel("Genre")

plt.show()
```

```python
country_counts = df['country'].value_counts().head(10)

plt.figure(figsize=(10,6))
plt.barh(country_counts.index[::-1], country_counts.values[::-1])

plt.title("Top 10 Countries by Number of Titles")
plt.xlabel("Number of Titles")
plt.ylabel("Country")

plt.show()
```

```python
language_counts = df['language'].value_counts().head(10)

plt.figure(figsize=(10,6))
plt.barh(language_counts.index[::-1], language_counts.values[::-1])

plt.title("Top 10 Content Languages")
plt.xlabel("Number of Titles")
plt.ylabel("Language")

plt.show()
```

```python
genre_rating = (
    df.groupby('genres')['vote_count']
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10,6))
plt.barh(genre_rating.index[::-1], genre_rating.values[::-1])

plt.title("Average Rating by Genre")
plt.xlabel("Average Rating")
plt.ylabel("Genre")

plt.show()
```

```python
plt.figure(figsize=(12,6))

plt.hist(df['vote_count'].dropna(), bins=20)

plt.title("Distribution of Content Ratings")
plt.xlabel("Vote Count")
plt.ylabel("Number of Titles")

plt.show()
```

```python
favorite_genres = df['genres'].value_counts()

plt.figure(figsize=(10,6))

plt.barh(
    favorite_genres.index[::-1],
    favorite_genres.values[::-1]
)

plt.title("Users' Favorite Genres")
plt.xlabel("Number of Users")
plt.ylabel("Favorite Genre")

plt.show()
```
